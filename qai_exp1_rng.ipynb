# Experiment 1: 16-Qubit Random Number Generator

### Aim:
Implement a 16-qubit Random Number Generator using Qiskit.

---
### Dataset Required:
None. Quantum measurement produces true non-deterministic randomness.

---
### Libraries Required & Purpose:
* `qiskit`: To create the quantum circuit and apply Hadamard and measurement gates.
* `qiskit_aer`: To simulate the quantum circuit locally.

---


### Step 1: Import Qiskit and Simulator

```python
# Import QuantumCircuit from qiskit to build the circuit
from qiskit import QuantumCircuit, transpile
# Import AerSimulator to execute the circuit locally
from qiskit_aer import AerSimulator
```

### Step 2: Build the 16-Qubit Circuit

```python
# Create a quantum circuit with 16 qubits and 16 classical bits
qc = QuantumCircuit(16, 16)

# Apply Hadamard (H) gate to each of the 16 qubits to create superposition
for i in range(16):
    qc.h(i)

# Measure all 16 qubits into the 16 classical bits
qc.measure(range(16), range(16))

# Display the circuit diagram
print(qc.draw(output='text'))
```

### Step 3: Run Simulation and Generate Random Number

```python
# Initialize the simulator backend
sim = AerSimulator()

# Transpile the circuit for the simulator
compiled_qc = transpile(qc, sim)

# Run the circuit for 1 shot to generate one random number
result = sim.run(compiled_qc, shots=1).result()
counts = result.get_counts()

# Get the generated 16-bit binary string
binary_number = list(counts.keys())[0]

# Convert the 16-bit binary number to a decimal integer (0 to 65535)
decimal_number = int(binary_number, 2)

print("Generated 16-bit Binary:", binary_number)
print("Generated Decimal Integer:", decimal_number)
```
