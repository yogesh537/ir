# Experiment 4: Quantum Teleportation Algorithm

### Aim:
Implement the Quantum Teleportation algorithm in Python using Qiskit.

---
### Dataset Required:
None. State vector is teleported via an entangled Bell pair.

---
### Libraries Required & Purpose:
* `qiskit`: To create Bell state entanglement, Bell measurement, and conditional gates.
* `qiskit_aer`: To simulate the teleportation process.

---


### Step 1: Import Qiskit and Simulator

```python
# Import QuantumCircuit, registers, and transpile
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
# Import AerSimulator
from qiskit_aer import AerSimulator
```

### Step 2: Build Quantum Teleportation Circuit

```python
# Qubits: q0 = state to teleport, q1 = Alice's Bell half, q2 = Bob's Bell half
qr = QuantumRegister(3, 'q')
cr_alice = ClassicalRegister(2, 'c_alice')
cr_bob = ClassicalRegister(1, 'c_bob')
qc = QuantumCircuit(qr, cr_alice, cr_bob)

# 1. Prepare state to teleport on q0 (let's prepare |1>)
qc.x(qr[0])

# 2. Create entangled Bell pair between q1 (Alice) and q2 (Bob)
qc.h(qr[1])
qc.cx(qr[1], qr[2])

# 3. Alice performs Bell-basis measurement on q0 and q1
qc.cx(qr[0], qr[1])
qc.h(qr[0])
qc.measure(qr[0], cr_alice[0])
qc.measure(qr[1], cr_alice[1])

# 4. Bob applies conditional corrections based on Alice's classical bits
qc.x(qr[2]).c_if(cr_alice[1], 1)
qc.z(qr[2]).c_if(cr_alice[0], 1)

# 5. Bob measures his qubit to verify teleportation
qc.measure(qr[2], cr_bob[0])

# Display circuit
print(qc.draw(output='text'))
```

### Step 3: Run Simulation

```python
# Run simulation
sim = AerSimulator()
compiled_qc = transpile(qc, sim)
counts = sim.run(compiled_qc, shots=1024).result().get_counts()

# Print results (Bob's qubit is always 1, matching the prepared state)
print("Simulation counts (Bob's bit | Alice's bits):", counts)
print("Teleportation verified: Bob receives state 1 in 100% of shots!")
```
