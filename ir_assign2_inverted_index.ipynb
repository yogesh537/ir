# Assignment 2: Document Retrieval using Inverted Files

### Aim:
Implement a program for retrieval of documents using inverted files.

---
### Dataset Required:
None. Uses the 2-document sample corpus provided in the lab manual.

---
### Libraries Required & Purpose:
* Standard Python built-ins (`dict`, `set`, `list`). No external heavy packages required.

---


### Step 1: Define Document Corpus

```python
# Define documents as given in lab manual
document1 = "The quick brown fox jumped over the lazy dog."
document2 = "The lazy dog slept in the sun."

print("Document 1:", document1)
print("Document 2:", document2)
```

### Step 2: Build the Inverted Index

```python
# Step 1: Tokenize and lowercase words
tokens1 = document1.lower().split()
tokens2 = document2.lower().split()

# Extract unique terms across all documents
terms = sorted(list(set(tokens1 + tokens2)))

# Step 2: Map each term to documents containing it
inverted_index = {}
for term in terms:
    documents = []
    if term in tokens1:
        documents.append("Document 1")
    if term in tokens2:
        documents.append("Document 2")
    inverted_index[term] = documents

# Step 3: Print the inverted index
print("=== Inverted Index ===")
for term, docs in inverted_index.items():
    print(term, "->", ", ".join(docs))
```

### Step 3: Query Document Retrieval

```python
# Function to search for a word
def search(query):
    return inverted_index.get(query.lower(), "Term not found")

# Test queries
print("Search 'lazy' :", search("lazy"))
print("Search 'fox'  :", search("fox"))
print("Search 'sun'  :", search("sun"))
```
