### Step 1: Import Libraries and Load Dataset

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler, normalize
from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
import scipy.cluster.hierarchy as shc

# Load CC_GENERAL.csv or create minimal fallback dataset
if os.path.exists('CC_GENERAL.csv'):
    X = pd.read_csv('CC_GENERAL.csv').drop('CUST_ID', axis=1, errors='ignore')
else:
    np.random.seed(42)
    n = 500
    X = pd.DataFrame({
        'BALANCE': np.random.exponential(1000, n),
        'PURCHASES': np.random.exponential(500, n),
        'CREDIT_LIMIT': np.random.choice([1000, 3000, 5000], n),
        'PAYMENTS': np.random.exponential(800, n)
    })

# Fill missing values
X = X.ffill().bfill()
```

### Step 2: Preprocess and Reduce Dimensionality to 2D using PCA

```python
# Scale and normalize features
X_scaled = StandardScaler().fit_transform(X)
X_norm = normalize(X_scaled)

# Apply PCA for 2D representation (P1, P2)
pca = PCA(n_components=2)
X_pca = pd.DataFrame(pca.fit_transform(X_norm), columns=['P1', 'P2'])
print("PCA features shape:", X_pca.shape)
```

### Step 3: Plot Dendrogram and Fit Agglomerative Clustering

```python
# Plot Dendrogram with Ward's linkage
plt.figure(figsize=(8, 4))
plt.title("Hierarchical Dendrogram (Ward Linkage)")
shc.dendrogram(shc.linkage(X_pca, method='ward'))
plt.show()

# Fit Agglomerative Clustering for optimal k = 2
ac2 = AgglomerativeClustering(n_clusters=2)
labels = ac2.fit_predict(X_pca)

# Visualize clusters
plt.figure(figsize=(6, 5))
plt.scatter(X_pca['P1'], X_pca['P2'], c=labels, cmap='rainbow', s=15)
plt.title("Agglomerative Clustering (k = 2)")
plt.show()

# Evaluate with Silhouette Score
score = silhouette_score(X_pca, labels)
print("Silhouette Score for k = 2:", round(score, 4))
```
