# Experiment 6: 5-Qubit Quantum Fourier Transform

### Aim:
Implement a 5-qubit Quantum Fourier Transform (QFT) using Qiskit.

---
### Dataset Required:
None. QFT transforms computational basis states into phase superpositions.

---
### Libraries Required & Purpose:
* `qiskit`: To build Hadamard, controlled-phase (cp), and SWAP gates.
* `numpy`: For rotation angles (pi / 2^k).
* `qiskit_aer`: To simulate the QFT circuit.

---


### Step 1: Import Qiskit, NumPy, and Simulator

```python
# Import QuantumCircuit, transpile, and AerSimulator
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
import numpy as np
```

### Step 2: Build 5-Qubit QFT Circuit

```python
# Create a 5-qubit circuit with 5 classical bits
qc = QuantumCircuit(5, 5)

# 1. State Preparation: Prepare test state |01101> (decimal 13)
qc.x(0)
qc.x(2)
qc.x(3)

# 2. Apply QFT (Hadamard + Controlled-Phase rotations)
for target in range(4, -1, -1):
    qc.h(target)
    for control in range(target - 1, -1, -1):
        angle = np.pi / (2 ** (target - control))
        qc.cp(angle, control, target)

# 3. Bit-reversal SWAP gates to restore standard qubit order
qc.swap(0, 4)
qc.swap(1, 3)

# 4. Measure all qubits
qc.measure(range(5), range(5))

# Display circuit layout
print(qc.draw(output='text'))
```

### Step 3: Run Simulation

```python
# Run the circuit on the simulator
sim = AerSimulator()
compiled_qc = transpile(qc, sim)
counts = sim.run(compiled_qc, shots=1024).result().get_counts()

# Print observed states (QFT produces equal superposition over all 32 basis states)
print(f"Total unique states observed in Fourier basis: {len(counts)} / 32")
print("Sample observed state counts:", dict(list(counts.items())[:5]))
```
