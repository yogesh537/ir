# Assignment 1: Text Pre-processing (Stop Words & Stemming)

### Aim:
Write a program for pre-processing of a text document such as stop word removal and stemming.

---
### Dataset Required:
None. Operates directly on sample sentences and a local 'text.txt' file.

---
### Libraries Required & Purpose:
* `nltk`: For text tokenization, standard stop words dictionary, and the Porter Stemmer.

---


### Step 1: Import NLTK and Download Corpora

```python
# Import required NLTK modules
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

# Download stopwords and tokenizers
nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
```

### Step 2: Remove Stop Words from a Sentence

```python
# Load English stop words
stop_words = set(stopwords.words('english'))

# Sample input sentence
sentence = "This is a sample sentence, showing off the stop words filtration."
tokens = word_tokenize(sentence)

# Filter out stop words
filtered = [w for w in tokens if w.lower() not in stop_words and w.isalnum()]

print("Original Tokens:", tokens)
print("Filtered Tokens:", filtered)
```

### Step 3: Stop Words Removal in a File

```python
# Create sample input file 'text.txt'
with open("text.txt", "w") as f:
    f.write("This is an example text file to demonstrate stop word removal in information retrieval.")

# Read file and remove stop words
with open("text.txt", "r") as f:
    words = f.read().split()

filtered_words = [w for w in words if w.lower() not in stop_words]

# Save output to 'filteredtext.txt'
with open("filteredtext.txt", "w") as f:
    f.write(" ".join(filtered_words))

print("Filtered file content saved to 'filteredtext.txt':", " ".join(filtered_words))
```

### Step 4: Stemming Words and Sentence using PorterStemmer

```python
# Initialize Porter Stemmer
ps = PorterStemmer()

# Stem a list of words
words = ["program", "programs", "programmer", "programming", "programmers"]
print("Word Stemming:")
for w in words:
    print(w, ":", ps.stem(w))

# Stem words in a full sentence
sentence = "Programmers program with programming languages"
words = word_tokenize(sentence)
print("\nSentence Stemming:")
for w in words:
    print(w, ":", ps.stem(w))
```
