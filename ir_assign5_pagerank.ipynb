### Step 1: Import NetworkX and Matplotlib

```python
import networkx as nx
import matplotlib.pyplot as plt
```

### Step 2: Implement PageRank on Directed Web Graph

```python
# Create directed web graph from manual (nodes: A, B, C, D)
G = nx.DiGraph()
G.add_edges_from([
    ('B', 'A'), ('B', 'C'),
    ('C', 'A'),
    ('D', 'A'), ('D', 'B'), ('D', 'C')
])

# Compute PageRank with damping factor alpha=0.85
pr = nx.pagerank(G, alpha=0.85)

print("=== PageRank Scores ===")
for node, rank in sorted(pr.items(), key=lambda x: x[1], reverse=True):
    print(f"Page {node}: {rank:.4f}")

# Visualize the graph
nx.draw(G, with_labels=True, node_color='lightcoral', node_size=1500, font_weight='bold', arrows=True)
plt.title("4-Page Directed Web Graph")
plt.show()
```

### Step 3: Barabási-Albert Scale-Free Graph (from manual page 33)

```python
# Generate Barabasi-Albert graph as shown on page 33 of manual
G_ba = nx.barabasi_albert_graph(60, 41, seed=42)

# Calculate PageRank with alpha=0.4
pr_ba = nx.pagerank(G_ba, 0.4)

# Display sample results
print("Sample PageRank values (First 5 nodes):")
for node in range(5):
    print(f"Node {node}: {pr_ba[node]:.6f}")
```
