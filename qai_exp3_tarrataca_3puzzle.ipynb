# Experiment 3: Tarrataca's 3-Puzzle Production System

### Aim:
Implement Tarrataca's quantum production system for the 3-puzzle problem.

---
### Dataset Required:
None. Puzzle state space is represented using 3 qubits.

---
### Libraries Required & Purpose:
* `qiskit`: To construct state transitions, superposition, and Grover search gates.
* `qiskit_aer`: To simulate amplitude amplification.

---


### Step 1: Import Qiskit and Simulator

```python
# Import QuantumCircuit and transpile from qiskit
from qiskit import QuantumCircuit, transpile
# Import AerSimulator
from qiskit_aer import AerSimulator
```

### Step 2: Build Tarrataca 3-Puzzle Circuit

```python
# Create a circuit with 3 qubits and 3 classical bits
qc = QuantumCircuit(3, 3)

# 1. Production Rules (tile transitions using CNOTs as shown in manual)
qc.cx(0, 1)
qc.cx(1, 2)

# 2. Put qubits into superposition using Hadamard gates
qc.h([0, 1, 2])

# 3. Grover Oracle for target solved state |111>
qc.h(2)
qc.ccx(0, 1, 2)
qc.h(2)

# 4. Grover Diffuser (Amplitude Amplification)
qc.h([0, 1, 2])
qc.x([0, 1, 2])
qc.h(2)
qc.ccx(0, 1, 2)
qc.h(2)
qc.x([0, 1, 2])
qc.h([0, 1, 2])

# 5. Measurement of all 3 qubits
qc.measure([0, 1, 2], [0, 1, 2])

# Display circuit
print(qc.draw(output='text'))
```

### Step 3: Run Simulation to Find Optimal State

```python
# Simulate the circuit
sim = AerSimulator()
compiled_qc = transpile(qc, sim)
counts = sim.run(compiled_qc, shots=1024).result().get_counts()

# Print results
print("Measurement counts:", counts)
best_state = max(counts, key=counts.get)
print("Optimal puzzle solution state found:", best_state)
```
